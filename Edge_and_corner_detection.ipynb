# **Edge and Corner Detection in Images using OpenCV(unit1)**

We used OpenCV sample images dataset

```python
!wget -q https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg
!wget -q https://raw.githubusercontent.com/opencv/opencv/master/samples/data/sudoku.png

print("✅ Dataset downloaded")
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("images/sudoku.png")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Gaussian Blur
blur = cv2.GaussianBlur(gray,(5,5),0)

# Canny Edge Detection
edges = cv2.Canny(blur,100,200)

# Harris Corner Detection
gray_float = cv2.cornerHarris(gray.astype('float32'),2,3,0.04)
dilate = cv2.dilate(gray_float, None)

corners = img.copy()
corners[dilate > 0.01 * dilate.max()] = [0,0,255]

# Show Results
titles = ["Original Image", "Gaussian Blur", "Canny Edges", "Harris Corners"]
images = [img, blur, edges, corners]

for i in range(4):
  plt.figure(figsize=(5,5))
  plt.title(titles[i])
  plt.imshow(images[i], cmap='gray')
  plt.axis('off')
```
